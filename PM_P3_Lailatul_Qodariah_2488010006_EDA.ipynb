## Latihan Mandiri
**Latihan 1.** Muat dataset lain bawaan seaborn, yaitu `tips` (`sns.load_dataset('tips')`). Tampilkan `head()`, `shape`, dan `info()`.


```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Memuat dataset tips
tips = sns.load_dataset('tips')

# Menampilkan 5 baris pertama
print("--- Tampilan 5 Baris Pertama ---")
print(tips.head())

# Menampilkan jumlah baris dan kolom
print("\n--- Dimensi Dataset ---")
print(tips.shape)

# Menampilkan informasi dataset
print("\n--- Informasi Dataset ---")
tips.info()

```

**Latihan 2.** Pada dataset `tips`, buat histogram kolom `total_bill` dan box plot kolom `tip`. Amati sebaran dan outlier-nya.


```python
# Histogram kolom total_bill
plt.figure(figsize=(8, 5))
sns.histplot(data=tips, x='total_bill', bins=15)
plt.title('Histogram Total Bill')
plt.xlabel('Total Bill')
plt.ylabel('Frekuensi')
plt.show()

# Box plot kolom tip
plt.figure(figsize=(8, 4))
sns.boxplot(data=tips, x='tip')
plt.title('Box Plot Tip')
plt.xlabel('Tip')
plt.show()
```

**Latihan 3.** Buat scatter plot `total_bill` (x) vs `tip` (y) pada dataset `tips`, diwarnai menurut kolom `time` (Lunch/Dinner). Apa yang dapat Anda simpulkan?


```python
# Scatter plot total_bill vs tip
sns.scatterplot(
    data=tips,
    x='total_bill',
    y='tip',
    hue='time'
)

plt.title('Scatter Plot Total Bill vs Tip')
plt.xlabel('Total Bill')
plt.ylabel('Tip')
plt.show()
```

**Latihan 4.** Buat heatmap korelasi untuk fitur numerik pada dataset `tips`. Pasangan fitur mana yang paling berkorelasi?


```python
# Memilih fitur numerik
num = tips.select_dtypes(include='number')

# Membuat heatmap korelasi
sns.heatmap(
    num.corr(),
    annot=True,
    cmap='coolwarm',
    fmt='.2f'
)

plt.title('Heatmap Korelasi Fitur Numerik')
plt.show()
```
